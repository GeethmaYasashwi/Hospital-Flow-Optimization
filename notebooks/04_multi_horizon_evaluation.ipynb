```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_absolute_error,
    root_mean_squared_error,
)

from forecasting import forecast_next_hours
```

```python
df = pd.read_csv(
    "../data/raw/ed_arrivals.csv"
)

df["timestamp"] = pd.to_datetime(
    df["timestamp"]
)

df = df.sort_values(
    "timestamp"
).reset_index(drop=True)

df.head()
```

```python
print("Number of rows:", len(df))

print(
    "Start:",
    df["timestamp"].min()
)

print(
    "End:",
    df["timestamp"].max()
)
```

```python
test_index = len(df) - 13

history = df.iloc[:test_index].copy()

actual_future = df.iloc[
    test_index:test_index + 12
].copy()

print("Last historical timestamp:")
print(history["timestamp"].iloc[-1])

print("\nActual future:")
print(actual_future[
    ["timestamp", "arrivals"]
])
```

```python
prediction = forecast_next_hours(
    history=history,
    hours=12,
)

prediction
```

```python
comparison = actual_future[
    ["timestamp", "arrivals"]
].merge(
    prediction,
    on="timestamp",
    how="inner"
)

comparison
```

```python
mae = mean_absolute_error(
    comparison["arrivals"],
    comparison["predicted_arrivals"]
)

print("12-hour forecast MAE:", mae)
```

```python
rmse = root_mean_squared_error(
    comparison["arrivals"],
    comparison["predicted_arrivals"]
)

print("12-hour forecast RMSE:", rmse)
```

```python
print(comparison)
print("MAE:", mae)
print("RMSE:", rmse)
```
